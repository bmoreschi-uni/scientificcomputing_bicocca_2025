## Q9: Poker odds
Use the deck of cards class from the notebook we worked through class to write a _Monte Carlo_ code that plays a lot of hands of straight poker (like 100,000).  Count how many of these hands has a particular poker hand (like 3-of-a-kind).  The ratio of # of hands with 3-of-a-kind to total hands is an approximation to the odds of getting a 3-of-a-kind in poker.

### Bonus: 
Just to practice modules, write that into a `.py` file to allow you to import and reuse them here.

````python
# this is the .py file with functions and classes that I need for the exercise
import poker
````

````python
hands = 100000
three = 0

for n in range(hands):
    # I create the deck, I shuffle it and I take 5 cards
    deck = poker.Deck()
    deck.shuffle()
    hand = deck.get_cards(5)

    if poker.three_of_a_kind(hand):
        three += 1

print("three of a kind is present {} times out of {}".format(three, hands))
````

````python
ratio = three / hands
print("odds of a three of a kind = ", ratio)
````

## Q12: Pascal's triangle

Pascal's triangle is created such that each layer has 1 more element than the previous, with `1`s on the side and in the interior, the numbers are the sum of the two above it, e.g.,:
```
            1
          1   1
        1   2   1
      1   3   3   1
    1   4   6   4   1
  1   5   10  10  5   1
```

1. Write a function to return the first `n` rows of Pascal's triangle.  The return should be a list of length `n`, with each element itself a list containing the numbers for that row.
2. Write a function to print out Pascal's triangle with proper formatting, so the numbers in each row are centered between the ones in the row above

````python
def triangle(n):
    
    rows = []

    # i is the number of rows
    for i in range(n):

        row = [1] # first number is 1

        for j in range(1, i): # 1 included, i excluded
            # if i = 0 and i = 1, the loop is skipped
            row.append(rows[i-1][j-1] + rows[i-1][j]) # previous row, sum the two numbers j-1 and j

        if i > 0:
            row.append(1) # last number of the row is 1

        rows.append(row)

    return rows



example = triangle(6)
print(example)
````

````python
def print_triangle(rows):

    appo = len(str(max(rows[-1]))) # the biggest number is in the last row, I need it to have an idea of the width of each box space 
    print("length of the biggest number is", appo)

    box = appo + 2 # space between numbers, one to the left, one to the right
    total = len(rows[-1]) * box # width of the last row to avoid overlap of numbers

    print(len(rows[-1]), "numbers in the last row, total width is", total)
    for row in rows:
        format_line = ""
        for x in row:
            format_line += "{:^{}}".format(x, box) # center each number in its box
        print(format_line.center(total)) # center respect the last row


print_triangle(example)
print_triangle(triangle(23)) # random n value to see if the format works
````
